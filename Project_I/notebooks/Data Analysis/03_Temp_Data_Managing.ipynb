| Column             | What it stores                                             | Relation to AQI Prediction                                          |
| ------------------ | ---------------------------------------------------------- | ------------------------------------------------------------------- |
| `datetime`         | Date/time of weather observation                           | Essential for time-series analysis and seasonal AQI patterns        |
| `tempmax`          | Maximum daily temperature (°C)                             | Indirectly affects pollutant dispersion and chemical reactions      |
| `tempmin`          | Minimum daily temperature (°C)                             | Important for winter inversion and pollution trapping               |
| `temp`             | Average daily temperature (°C)                             | Strong indirect influence on AQI behavior                           |
| `dew`              | Dew point temperature (°C) indicating atmospheric moisture | Indirectly affects particulate formation and fog/smog               |
| `humidity`         | Relative humidity (%)                                      | Strong indirect effect; high humidity can increase PM concentration |
| `precipprob`       | Probability of precipitation (%)                           | Rain can reduce air pollutants by washing particles                 |
| `precipcover`      | Percentage of area/time covered by precipitation           | Indirect effect on pollutant removal                                |
| `preciptype`       | Type of precipitation (rain, snow, etc.)                   | Minor indirect influence on AQI                                     |
| `windgust`         | Maximum short-duration wind speed                          | Helps disperse pollutants                                           |
| `windspeed`        | Average wind speed                                         | Very important; strong winds reduce pollution concentration         |
| `winddir`          | Wind direction (degrees)                                   | Influences pollutant transport sources                              |
| `sealevelpressure` | Atmospheric pressure adjusted to sea level                 | Important for weather stability and pollution accumulation          |
| `cloudcover`       | Percentage of sky covered by clouds                        | Indirect influence on temperature and atmospheric mixing            |
| `visibility`       | Horizontal visibility distance                             | Often strongly related to particulate pollution and haze            |
| `solarradiation`   | Solar radiation received (W/m²)                            | Affects photochemical pollution formation                           |
| `solarenergy`      | Total solar energy received                                | Similar indirect effect on atmospheric chemistry                    |
| `uvindex`          | Ultraviolet radiation index                                | Influences ozone and photochemical reactions                        |
| `severerisk`       | Risk index for severe weather events                       | Usually weak relation to AQI prediction                             |
| `sunrise`          | Daily sunrise time                                         | Useful only for advanced temporal feature extraction                |
| `sunset`           | Daily sunset time                                          | Minor relevance unless modeling diurnal patterns                    |
| `precipitation`    | Actual precipitation/rainfall amount                       | Very important; rainfall reduces airborne particles                 |
